#### **1. Обоснование необходимости разработки**


Для кого предназначено разрабатываемое Вами решение и какие потребности целевой аудитории оно призвано удовлетворить?




*Обоснуйте актуальность решаемой проблемы для конкретных заинтересованных сторон.*

Инструмент ориентирован на органы и организации, осуществляющие экологический контроль и градостроительное планирование в Москве, а также на специалистов, участвующих в проектировании объектов программы реновации.

Реализация программы реновации затрагивает лесные территории города, и её последствия для зелёного фонда требуют систематической, количественной и воспроизводимой оценки — как на этапе проектирования объектов, так и при последующем контроле. Единый программный инструмент позволяет проводить такую оценку быстро, объективно и сопоставимо между разными объектами реновации, что особенно важно при большом числе проектов в разных районах города.

#### **2. Ожидаемые результаты проекта**



Опишите конкретные результаты, которые планируется получить по завершении проекта. Какой программный продукт, методика или информационная система будет создан/создана в результате вашей работы?

Результатом проекта является программный инструмент, реализующий разрабатываемую методику оценки воздействия программы реновации на лесные территории. Инструмент совмещает два вида анализа: пространственный — выявление пересечений и буферных зон между объектами реновации и лесными территориями с расчётом показателей текущего воздействия, и временной — анализ динамики состояния лесного покрова по разновременным спутниковым снимкам за предшествующий период, на основе которого формируется прогнозная оценка ожидаемого воздействия реновации. Результаты представляются в виде картографической визуализации и табличного отчёта, включающего как текущие показатели, так и прогнозные оценки.

Технически инструмент реализуется как плагин для QGIS — это позволяет встроить методику непосредственно в среду, где уже ведётся работа с пространственными данными, и обеспечить воспроизводимость оценки при появлении новых объектов программы реновации.

#### **3. Методология использования**


Каким образом предполагается практическое применение разрабатываемого решения? Опишите основные сценарии взаимодействия пользователей с разрабатываемым решением.

Пользователь подключает в среде QGIS исходные пространственные слои — границы объектов программы реновации и границы лесных территорий — а также временной ряд спутниковых снимков за период, предшествующий началу работ на рассматриваемой территории. Плагин выполняет два связанных этапа анализа: во-первых, определяет пространственные пересечения и буферные зоны между объектами реновации и лесными территориями; во-вторых, анализирует динамику состояния лесного покрова за предшествующий период (например, по вегетационным индексам) для выявления сложившегося тренда. На основе сочетания текущего пространственного анализа и выявленной динамики формируется прогнозная оценка ожидаемого воздействия. Результат представляется в двух формах: картографической — зоны воздействия и их прогнозные категории, и табличной — отчёт с текущими и прогнозными показателями по каждому объекту.

#### **4. Входные данные и их источники**


Какие типы данных будут использоваться в качестве входных данных для вашего решения? В каком формате они будут представлены и из каких источников планируется их получение?

Предполагается использование трёх групп данных. Первая — границы объектов программы реновации (открытые данные Правительства Москвы либо материалы Мосгоргеотреста). Вторая — границы лесных и иных зелёных территорий города (реестр ООПТ, данные Департамента природопользования и охраны окружающей среды г. Москвы, либо OpenStreetMap). Третья — разновременные спутниковые снимки (например, серия сцен Sentinel-2) за период, предшествующий реализации программы на рассматриваемой территории, необходимые для анализа динамики лесного покрова и построения прогноза. Точный состав источников будет уточнён на этапе анализа их доступности.

#### **5. Выходные данные и практическая значимость**


Какие результаты будет формировать Ваше решение на выходе?

На выходе инструмент формирует картографическую визуализацию зон воздействия и табличный отчёт, включающий как показатели текущего пространственного пересечения объектов реновации с лесными территориями, так и прогнозную оценку ожидаемого воздействия, построенную на основе анализа динамики лесного покрова за предшествующий период. Практическая значимость: инструмент даёт возможность оценивать воздействие не только по факту, но и заранее, на этапе планирования — это позволяет учитывать прогноз при корректировке границ застройки, планировании компенсационных мероприятий и контроле соблюдения экологических требований при реализации программы реновации.
