# **Руководство по формированию темы дипломного проекта на основе анализа научных публикаций**

---

## **1. Введение**

Настоящее руководство описывает методику формирования темы дипломного проекта на основе анализа актуальных научных публикаций. Руководство предназначено для студентов, обучающихся по направлению, связанному с обработкой пространственных данных, и выполняющих задания курса «Управление проектами информационных систем».

Курс построен таким образом, что каждое задание формирует отдельный раздел дипломной работы. Задания выстроены в соответствии с фазами жизненного цикла проекта:



| №  | Задание | Фаза проекта |
|----|---------|--------------|
| 1  | Определение концептуальных основ дипломного проекта | Инициация |
| 2  | Сравнительный анализ существующих решений (анализ аналогов) | Анализ |
| 3  | Формирование иерархической структуры проекта (WBS) | Анализ и структурирование |
| 4  | Разработка спецификации требований к ИС (SRS) | Требования |
| 5  | Формирование календарного плана проекта и управление рисками | Планирование |
| 6  | Разработка UML-диаграмм для проектирования ИС | Проектирование |
| 7  | Проектирование базы данных для ИС | Проектирование |
| 8  | Разработка пользовательских интерфейсов (прототипы) | Проектирование |
| 9  | Разработка стратегии тестирования и тест-кейсов | Верификация |
| 10 | Разработка презентации дипломного проекта | Завершение |



Для корректного выполнения заданий необходимо определить тему дипломного проекта. Данное руководство содержит пошаговую методику выбора темы с использованием научных публикаций в качестве отправной точки.

## **3. Принципы формирования темы на основе научных публикаций**

### **3.1. Принцип преемственности**

В научной и инженерной практике подавляющее большинство работ строится на результатах предшествующих исследований. Развитие архитектур глубокого обучения наглядно иллюстрирует данный принцип:



```
AlexNet (2012)
    → VGG (2014) — увеличение глубины сети
        → ResNet (2015) — введение остаточных связей
            → Transformer (2017) — механизм внимания
                → Vision Transformer (2020) — перенос трансформеров на задачи компьютерного зрения
                    → Segment Anything (2023) — универсальная модель сегментации
```



Каждая последующая работа опирается на предыдущую и вносит конкретное улучшение. Дипломный проект может строиться по аналогичному принципу.

### **3.2. Базовая формула**

Формирование темы дипломного проекта может основываться на следующей зависимости:

> **Тема диплома = Задача из научной публикации (или нескольких публикаций) + Собственный вклад (модификация, улучшение, новый контекст применения)**



При этом в качестве основы может выступать как одна публикация, так и несколько работ по одной предметной области. Например, студент может проанализировать 3–5 статей по задаче сегментации спутниковых снимков, в которых рассмотрены различные методы (U-Net, DeepLab, SegFormer), и на основе их сравнительного анализа сформулировать собственную тему.

### **3.3. Подходы к использованию публикаций**

| Подход | Описание | Когда применять |
|--------|----------|-----------------|
| **Одна базовая статья** | Берётся конкретная публикация, <br> к результатам которой применяется <br> одна из стратегий улучшения <br> (раздел 4) | Когда найдена статья <br> с чётким и воспроизводимым методом, <br> к которому можно внести <br> конкретное улучшение |
| **Несколько статей <br> по одной задаче** | Анализируются 3–5 публикаций, <br> решающих одну и ту же задачу <br> разными методами. <br> На основе анализа формулируется тема, <br> объединяющая или сравнивающая <br> подходы | Когда интересна сама задача <br> (например, сегментация, <br> детекция изменений), <br> но нет привязки <br> к конкретному методу |
| **Несколько статей <br> по одной области** | Анализируются публикации <br> из одной предметной области <br> (например, мониторинг <br> сельского хозяйства), <br> но с разными задачами и методами. <br> Тема формируется <br> как комплексное решение | Когда интересна <br> прикладная область, <br> а не конкретная <br> техническая задача |



В каждом из этих подходов найденные публикации впоследствии могут использоваться как аналоги в Задании №2 (сравнительный анализ существующих решений).

## **4. Стратегии формирования темы**

Ниже описаны семь стратегий, позволяющих сформулировать тему дипломного проекта на основе существующих публикаций. Стратегии могут применяться по отдельности или комбинироваться.

### **Стратегия 1. Применение к другому датасету или региону**

**Описание.** Авторы публикации обучали и тестировали модель на данных определённого региона или на конкретном датасете. Студент применяет тот же или модифицированный метод к данным другого региона или другого набора данных.

**Обоснование.** Модели, обученные на данных из Европы или Северной Америки, могут демонстрировать иные показатели качества на данных из Центральной Азии вследствие различий в ландшафте, застройке, климатических условиях и качестве доступных снимков. Адаптация метода к новому региону является самостоятельным научным и инженерным вкладом.

**Пример.**
- Исходная публикация: сегментация зданий по спутниковым снимкам городов Германии на основе U-Net
- Тема диплома: «Разработка системы автоматической сегментации зданий по спутниковым снимкам городов Казахстана на основе архитектуры U-Net»

---

### **Стратегия 2. Использование другой архитектуры или метода**

**Описание.** Авторы публикации использовали определённую модель или архитектуру. Студент решает ту же задачу с использованием другого метода, который ранее не использовался, и проводит сравнительный анализ результатов. При анализе нескольких публикаций студент может выбрать наиболее перспективный метод или предложить модификацию одного из рассмотренных подходов.

**Обоснование.** Область глубокого обучения развивается быстро. Публикации 2021–2023 годов могут не учитывать архитектуры, появившиеся позднее (например, модели на основе State Space Models, такие как Mamba). Сравнение современных и ранее предложенных методов на одной задаче позволяет установить их относительную эффективность.

**Пример.**
- Исходные публикации: 3 статьи по классификации земного покрова, использующие CNN, Random Forest и Vision Transformer соответственно
- Тема диплома: «Сравнительный анализ методов классификации земного покрова по мультиспектральным данным Sentinel-2 на основе свёрточных нейронных сетей и архитектуры Vision Transformer»

---

### **Стратегия 3. Добавление типа данных (мультимодальный подход)**

**Описание.** Публикация использует один тип данных (например, оптические спутниковые снимки). Студент добавляет данные другого типа — данные радиолокационной съёмки SAR, данные LiDAR, мультиспектральные каналы, данные о рельефе (DEM) и т.д.

**Обоснование.** Различные типы данных содержат комплементарную информацию. Оптические снимки непригодны при облачности, тогда как SAR-данные не зависят от погодных условий. LiDAR обеспечивает информацию о высоте объектов, отсутствующую в двумерных снимках. Комбинирование нескольких источников данных (мультимодальный подход) является одним из актуальных направлений в обработке пространственных данных.

**Пример.**
- Исходная публикация: детекция затоплений по оптическим спутниковым снимкам
- Тема диплома: «Разработка мультимодальной системы детекции затоплений на основе совместного анализа оптических данных Sentinel-2 и радарных данных Sentinel-1»

---

### **Стратегия 4. Оптимизация и упрощение модели**

**Описание.** Авторы публикации создали ресурсоёмкую модель, требующую значительных вычислительных мощностей. Студент разрабатывает облегчённую версию, сохраняя приемлемое качество.

**Обоснование.** В условиях реальной эксплуатации — на беспилотных летательных аппаратах (unmanned aerial vehicles, UAV), мобильных устройствах (mobile devices), в полевых условиях (field conditions) — доступ к высокопроизводительным графическим процессорам (GPU, Graphics Processing Unit) ограничен. Разработка компактных и быстрых моделей с использованием методов дистилляции знаний (knowledge distillation), прореживания (pruning) и квантования (quantization) представляет собой самостоятельную инженерную задачу.


**Пример.**
- Исходная публикация: семантическая сегментация аэрофотоснимков на основе Swin Transformer (модель с 200 млн параметров)
- Тема диплома: «Разработка облегчённой модели семантической сегментации аэрофотоснимков для развёртывания на устройствах с ограниченными вычислительными ресурсами»

---

### **Стратегия 5. Перенос метода в новую прикладную область**

**Описание.** Метод из публикации применяется для одной задачи. Студент переносит его в другую предметную область, где аналогичная задача также актуальна, но ранее не решалась данным способом.

**Обоснование.** Методы, показавшие высокую эффективность в одной области, нередко могут быть адаптированы для смежных задач (transfer learning). Перенос метода в новую область с учётом её специфики является полноценным вкладом.

**Пример.**
- Исходная публикация: детекция автомобилей на аэрофотоснимках для мониторинга трафика
- Тема диплома: «Применение методов детекции объектов на аэрофотоснимках для автоматического мониторинга состояния сельскохозяйственных угодий»

---

### **Стратегия 6. Сравнительный анализ и разработка гибридного метода**

**Описание.** Студент анализирует несколько публикаций, в которых одна и та же задача решается различными методами. На основе сравнительного анализа выявляются сильные и слабые стороны каждого подхода. Результатом может стать ансамблевый или гибридный метод, объединяющий преимущества рассмотренных подходов.

**Обоснование.** В большинстве практических задач не существует единственного «лучшего» метода. Различные подходы демонстрируют преимущества в различных условиях. Систематическое сравнение и обоснованное комбинирование методов является инженерно значимой задачей.

**Пример.**
- Исходные публикации: статьи по классификации облаков точек с использованием PointNet++, DGCNN и RandLA-Net
- Тема диплома: «Сравнительный анализ и разработка гибридного метода классификации облаков точек LiDAR для 3D-моделирования городской среды»

---

### **Стратегия 7. Расширение пайплайна и разработка информационной системы**

**Описание.** Публикация описывает модель или алгоритм, но не рассматривает вопросы его практического внедрения. Студент добавляет недостающие этапы: предобработку данных, постобработку результатов, визуализацию, веб-интерфейс, интеграцию в ГИС, разработку API.

**Обоснование.** Данная стратегия особенно актуальна для дипломных проектов по направлению «информационные системы». Дипломный проект предполагает разработку не отдельной модели, а системы в целом. Превращение научного результата в работающий программный продукт является самостоятельной и технически сложной задачей.

**Пример.**
- Исходные публикации: несколько статей по нейросетевой классификации земного покрова
- Тема диплома: «Разработка веб-ГИС для автоматической классификации земного покрова на основе нейросетевого анализа мультиспектральных спутниковых данных»

---

### **Сводная таблица стратегий**

| № | Стратегия | Основной вопрос | Рекомендуемое количество базовых публикаций |
|---|-----------|-----------------|---------------------------------------------|
| 1 | Другой датасет / регион | «Применялся ли данный метод к данным интересующего региона?» | 1–2 |
| 2 | Другая архитектура / метод | «Существуют ли более современные или альтернативные методы решения данной задачи?» | 2–5 |
| 3 | Добавить тип данных | «Какие дополнительные источники данных могут повысить качество решения?» | 1–3 |
| 4 | Оптимизировать модель | «Возможно ли обеспечить сопоставимое качество при меньших вычислительных затратах?» | 1–2 |
| 5 | Новая прикладная область | «В каких смежных областях данный метод может быть востребован?» | 1–3 |
| 6 | Сравнение и гибрид | «Какие методы решения данной задачи существуют и можно ли их объединить?» | 3–5 |

Стратегии желательно комбинировать.

Например, сочетание стратегии 1 (другой регион) + стратегии 2 (другая архитектура) + стратегии 7 (веб-интерфейс) позволяет сформировать комплексную тему дипломного проекта.

## **6. Методика экспресс-анализа научной статьи**

На этапе поиска темы нет необходимости читать каждую публикацию полностью. Ниже приведён алгоритм экспресс-анализа, позволяющий оценить релевантность статьи за 5–7 минут.

### **Шаг 1. Заголовок (10–15 секунд)**

Оцените название публикации по следующим критериям:
- Понятна ли решаемая задача?
- Относится ли работа к обработке пространственных данных?
- Представляет ли тема интерес для дальнейшего развития?

### **Шаг 2. Аннотация / Abstract (1–2 минуты)**

Аннотация содержит краткое описание работы в 10–15 строках. После прочтения аннотации необходимо уметь ответить на три вопроса:
- Какая задача решается?
- Какой метод предложен?
- Какой результат получен (метрики качества)?

### **Шаг 3. Рисунки и таблицы (2–3 минуты)**

Просмотрите публикацию, обращая внимание только на иллюстрации и таблицы:
- Рисунки, как правило, отображают архитектуру модели, примеры входных и выходных данных, визуализацию результатов
- Таблицы содержат сравнение с другими методами, характеристики набора данных, количественные результаты экспериментов

По рисункам и таблицам можно составить представление о содержании примерно 70% публикации.

### **Шаг 4. Заключение / Conclusion (1–2 минуты)**

## **9. Типичные ошибки при выборе темы**

### **Ошибка 1. Слишком широкая формулировка**

**Некорректно:** «Применение искусственного интеллекта для обработки пространственных данных»

**Причина:** формулировка охватывает целую научную дисциплину и не может быть раскрыта в рамках одной дипломной работы.

**Корректно:** «Применение архитектуры SegFormer для сегментации сельскохозяйственных угодий по данным Sentinel-2 в Алматинской области»

**Рекомендация:** в формулировке темы должны быть конкретизированы задача, метод и предметная область (или данные).

---

### **Ошибка 2. Отсутствие собственного вклада**

**Некорректно:** «Запуск предобученной модели YOLOv8 на датасете DOTA»

**Причина:** работа сводится к воспроизведению существующего решения без модификации. Это не соответствует требованиям к дипломному проекту.

**Корректно:** «Сравнительный анализ моделей семейства YOLO для детекции объектов инфраструктуры на аэрофотоснимках с разработкой веб-интерфейса визуализации результатов»

**Рекомендация:** в теме должен быть явно обозначен элемент, составляющий вклад автора — модификация метода, новые данные, системная интеграция, сравнительный анализ.

---

### **Ошибка 3. Отсутствие доступных данных**

**Некорректно:** «Классификация ультравысокодетальных снимков со спутника WorldView-4 с разрешением 30 см»

**Причина:** коммерческие спутниковые данные имеют высокую стоимость. При отсутствии доступа к данным выполнение проекта невозможно.

**Корректно:** «Классификация земного покрова по открытым данным Sentinel-2 с использованием методов глубокого обучения»

**Рекомендация:** на этапе выбора темы необходимо убедиться в доступности данных. Рекомендуется использовать открытые источники: Sentinel-1/2 (Copernicus), Landsat (USGS), Google Earth Engine, открытые датасеты на Kaggle.

---

### **Ошибка 4. Отсутствие системной составляющей**

**Некорректно:** «Исследование влияния функции потерь на точность семантической сегментации»

**Причина:** формулировка описывает чисто исследовательскую задачу без системной составляющей. Дипломный проект по направлению «информационные системы» предполагает разработку системы, а не только модели.

**Корректно:** «Разработка информационной системы семантической сегментации спутниковых снимков с исследованием влияния функции потерь на точность результатов»

**Рекомендация:** тема должна содержать указание на разработку системы (стратегия 7). Модель или алгоритм выступают частью системы, а не самостоятельным результатом.

---

### **Ошибка 5. Избыточная фиксация на первоначальной формулировке**

Тема дипломного проекта уточняется в процессе работы. На текущем этапе достаточно определить направление исследования. Точная формулировка корректируется по мере выполнения заданий курса — в первую очередь после анализа аналогов (Задание №2) и разработки спецификации требований (Задание №4). Если в процессе работы возникает необходимость скорректировать тему, следует обсудить изменения с преподавателем.
