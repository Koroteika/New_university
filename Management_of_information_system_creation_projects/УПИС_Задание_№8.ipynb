# **Задание №8. Разработка пользовательских интерфейсов (прототипы интерфейса)**

## **Введение**

Проектирование пользовательского интерфейса является важным этапом в разработке информационных систем, определяющим удобство использования продукта и удовлетворённость конечных пользователей. В контексте управления проектами создания информационных систем качественное проектирование интерфейса существенно снижает затраты на доработки и поддержку, минимизирует количество ошибок пользователей и повышает эффективность работы с системой.




**Качественное проектирование интерфейса позволяет:**
- Снизить время обучения пользователей на 50-80%
- Уменьшить количество ошибок пользователей на 60-90%
- Повысить производительность работы с системой на 20-50%
- Сократить затраты на техническую поддержку на 30-40%
- Увеличить удовлетворённость пользователей и лояльность к продукту

В данном задании вы разработаете визуальные прототипы интерфейса вашей информационной системы, которые станут основой для реализации пользовательского взаимодействия.

## **Связь с предыдущими заданиями**

#### **Теоретическая часть:**

**Информационная архитектура** — это структурная организация контента и функциональности системы, обеспечивающая интуитивную навигацию и эффективный поиск информации.

**Основные компоненты:**

- **Карта навигации (структура сайта/приложения):**
  - Иерархическая структура всех разделов
  - Связи между разделами
  - Глубина вложенности (рекомендуется не более 3-4 уровней)

- **Пользовательские пути:**
  - Последовательность действий для достижения цели
  - Точки принятия решений
  - Альтернативные маршруты

### **2. Каркасные модели (схематичные прототипы)**

**Каркасная модель** (англ. wireframe) — это схематичное представление структуры интерфейса без визуального оформления, фокусирующееся на расположении элементов и функциональности.



![Текст ссылки](https://visme.co/blog/wp-content/uploads/2022/01/mobile-ui-design-wireframe-example-768x571.jpg)

**Характеристики каркасных моделей:**
- Монохромное исполнение (чёрно-белое или оттенки серого)
- Простые геометрические формы
- Условные обозначения для контента
- Акцент на структуре, а не на визуальном стиле



**Элементы каркасной модели:**
- Блоки контента (прямоугольники с подписями)
- Навигационные элементы
- Формы ввода
- Кнопки действий
- Заглушки для изображений

### **3. Детальные макеты (визуальные прототипы)**

**Детальный макет** (англ. mockup) — это высокодетализированный визуальный прототип, максимально приближенный к финальному дизайну продукта.

**Компоненты детального макета:**
- **Цветовая схема:** основные и вспомогательные цвета
- **Типографика:** шрифты, размеры, межстрочные интервалы
- **Сетка:** модульная система расположения элементов
- **Компоненты интерфейса:** кнопки, поля ввода, карточки
- **Иконки и иллюстрации**
- **Реальный контент** (вместо заглушек)

### **4. Принципы проектирования интерфейсов**

#### **4.1. Принципы юзабилити Якоба Нильсена**

1. **Видимость состояния системы** — пользователь должен понимать, что происходит
2. **Соответствие системы реальному миру** — использование понятных терминов
3. **Свобода действий** — возможность отмены и повтора
4. **Согласованность и стандарты** — единообразие элементов
5. **Предотвращение ошибок** — проектирование, минимизирующее ошибки
6. **Узнавание вместо вспоминания** — видимость опций и действий
7. **Гибкость и эффективность** — ускорители для опытных пользователей
8. **Эстетичный и минималистичный дизайн** — отсутствие лишней информации
9. **Помощь в распознавании и исправлении ошибок** — понятные сообщения
10. **Справка и документация** — доступная помощь при необходимости

#### **4.2. Законы проектирования интерфейсов**

**Закон Фиттса:**
Время достижения цели зависит от расстояния до цели и её размера. Важные элементы должны быть крупными и легкодоступными.

**Закон Хика:**
Время принятия решения увеличивается с ростом количества альтернатив. Минимизируйте выбор.

**Закон Миллера:**
Человек может удерживать в краткосрочной памяти 7±2 элемента. Группируйте информацию.

**Принцип Парето:**
80% пользователей используют 20% функций. Приоритизируйте основной функционал.

### **5. Особенности проектирования различных типов интерфейсов**

#### **5.1. Графический интерфейс пользователя**

**Веб-интерфейс:**
- Адаптивная вёрстка для различных разрешений
- Кроссбраузерная совместимость
- Оптимизация загрузки

**Мобильный интерфейс:**
- Сенсорное управление (размер касания минимум 44×44px для iOS, 48×48dp для Android)
- Жесты и свайпы
- Ориентация экрана

**Десктопное приложение:**
- Многооконный интерфейс
- Контекстные меню
- Горячие клавиши

#### **5.2. Интерфейс командной строки (консольный интерфейс)**

**Структура команд:**



```
команда [опции] [аргументы]

Пример:
analyze --input photo.jpg --output result.txt --verbose
```



**Принципы проектирования консольного интерфейса:**
- Согласованность синтаксиса команд
- Информативные сообщения об ошибках
- Подробная справка (--help)
- Автодополнение команд
- Цветовое выделение вывода

**Пример структуры вывода:**



```
[ИНФО] Начало анализа изображения...
```

### **1. ИНФОРМАЦИОННАЯ АРХИТЕКТУРА**

#### **1.1. Карта навигации (структура системы)**



```
Главный экран
├── Найти ингредиент
│   └── Страница ингредиента
│       └── Добавить в список
├── Анализ состава
│   ├── Выбор источника
│   │   ├── Камера
│   │   └── Галерея
│   └── Результаты анализа
│       ├── Вкладка "Анализ" (список ингредиентов)
│       ├── Вкладка "Текст" (редактирование)
│       └── Вкладка "Изображение"
└── Боковое меню
    ├── Выбор пользователя
    ├── Настройки пользователя
    │   ├── Вкладка "Особенности"
    │   └── Вкладка "Мои ингредиенты"
    └── Добавить пользователя
```



#### **1.2. Основные пользовательские пути**

**Сценарий 1: Анализ состава продукта**



```
Главный экран → Кнопка "Анализ состава" → Выбор источника (Камера/Галерея) →
→ Захват/выбор изображения → Обработка изображения → Распознавание текста →
→ Отображение результатов (список ингредиентов с предупреждениями)

```



**Сценарий 2: Поиск конкретного ингредиента**



```
Главный экран → Кнопка "Найти ингредиент" → Ввод названия в поиск →
→ Фильтрация списка → Выбор ингредиента → Просмотр информации →
→ Добавление в пользовательский список (опционально)

```



**Сценарий 3: Настройка профиля пользователя**



```
Главный экран → Боковое меню → Настройки пользователя →
→ Выбор пищевых привычек → Сохранение изменений

```

Шаг 1: выполнить анализ для объекта → Шаг 2: система не находит пересечения с лесными территориями → Шаг 3: формируется короткий отчёт «Воздействие не обнаружено» без блока прогноза → Шаг 4: запись сохраняется в историю со статусом «Нет пересечения»

**Сценарий 4: Пакетный/программный запуск через Processing Toolbox**

Шаг 1: аналитик открывает алгоритм методики в Processing Toolbox, минуя dock-панель → Шаг 2: заполняет форму параметров (объект, буфер, период) → Шаг 3: запускает Run → Шаг 4: получает выходной отчёт/слой, пригодный для использования в собственных скриптах или при пакетной обработке нескольких объектов

## **5. Адаптация под различные интерфейсы**

Проектируемая система — desktop-плагин QGIS, а не мобильное приложение и не консольная утилита, поэтому пункт «адаптация» интерпретирован в контексте, применимом к desktop-GIS-инструменту, а не подменён неприменимыми здесь экранами телефона.

### **5.1. Узкая закреплённая панель vs широкое плавающее окно**

image.png

QGIS позволяет закреплять (docking) панели по краю окна в узком виде или откреплять их в свободно перемещаемое и произвольно расширяемое окно. В узком варианте поля идут одной колонкой сверху вниз; в широком — перекомпонованы в две колонки (поля слева, чекбокс/предупреждение/кнопка запуска справа), при этом набор элементов и их порядок в потоке остаются идентичными — это и есть responsive-эквивалент для desktop-плагина.

### **5.2. Альтернативный вход — диалог Processing Toolbox**

Роль, которую для веб/мобильных систем обычно играет консольный интерфейс, здесь выполняет стандартный диалог параметров QGIS Processing Toolbox (см. каркасную модель 2.5): минимальная, автоматически генерируемая QGIS форма «параметр — значение» без произвольного оформления, пригодная для пакетного запуска через Python-консоль QGIS или собственные скрипты аналитика — то есть программный, а не только интерактивный доступ к методике, как и требует API-часть интерфейсных требований из Задания №4.

### **6.2. Применение законов проектирования**

**Закон Фиттса:** кнопка «Выполнить анализ» растянута на всю ширину панели и имеет высоту 48px — это самая крупная кликабельная область на экране, так как это основное и самое частое действие аналитика.

**Закон Хика:** экран «Новый анализ» ограничен четырьмя параметрами вместо десятков теоретически возможных ГИС-настроек — сознательно сокращённый набор выборов, напрямую отражающий узкий scope методики (Задание №4: анализ одного объекта, а не всего города).

**Закон Миллера:** 2 вкладки в dock-панели и 4 раздела отчёта — все группы элементов далеко укладываются в пределы 7±2, минимизируя когнитивную нагрузку.

### **6.3. Улучшения относительно аналогов**

**Проблема в аналогах (Задание №2):** ни один из рассмотренных аналогов (Environmental Assessment Studio, URGREECS, QGYF, i-Tree Canopy) не сочетает привязку к конкретному объекту застройки, анализ временной динамики и прогнозирование, а разрыв между автоматизированным ГИС-анализом и официальной методикой №897-ПП нигде не показан явно.

- Нет привязки к конкретному объекту застройки → форма «Новый анализ» требует выбрать конкретный объект реновации, а не произвольную область на карте.
- Нет прогнозной составляющей → отдельный блок «Прогноз воздействия» на экранах результатов и отчёта.
- Разрыв между ГИС-анализом и официальной методикой → прямая текстовая отсылка к методике №897-ПП в интерфейсе отчёта, явно обозначающая место инструмента в общем процессе.
- Нет локализации под РФ/Москву → интерфейс полностью на русском языке, примеры объектов — реальные районы Москвы.

## **7. Спецификация интерфейса**

### **7.1. Сетка и отступы**

- Базовая сетка: 8px (соответствует конвенции Qt/QGIS)
- Внутренние отступы содержимого панели: 24px по горизонтали
- Вертикальный ритм между смысловыми блоками: 16–18px
- Высота поля ввода/кнопки: 40–48px, радиус скругления: 6–8px

### **7.2. Компоненты интерфейса**

**Кнопки:** первичная — залитая зелёным (#2E7D32), белый жирный текст, высота 48px; вторичная — белый фон, зелёная обводка 2px, зелёный текст.

**Поля ввода/выпадающие списки:** белый фон, серая обводка 1px (#B0B0B0), радиус 6px, высота 40px; подпись заглавными буквами над полем, 12–13px, серый (#6B6B6B).

**Чекбокс:** 22×22px, залит зелёным при выборе, белая галочка.

**Карточки результатов:** светло-зелёный фон (#F5F9F5), радиус 8px, подпись заглавными буквами + крупное значение (18px, жирный).

**Таблица (история анализов):** чередующаяся заливка строк, серый заголовок, левое выравнивание текста.

### **7.3. Анимации и переходы**

Плагин — часть настольного приложения QGIS, а не веб/мобильный продукт, поэтому анимации минимальны и функциональны: плавное появление экрана результатов после завершения анализа (150–200ms), мгновенное переключение активной вкладки без анимации, как принято в Qt-интерфейсах, без декоративных переходов, отвлекающих аналитика от рабочего процесса.
