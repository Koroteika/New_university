# **Задание №2. Сравнительный анализ существующих решений (анализ аналогов)**

## **Введение**

**Анализ существующих решений является важным этапом** в управлении проектами создания информационных систем.

Прежде чем приступить к разработке нового программного продукта, необходимо изучить рынок, понять, какие решения уже существуют, оценить их сильные и слабые стороны, а также определить, какую уникальную ценность может принести ваш проект.

Согласно исследованиям, более 60% неудачных IT-проектов сталкиваются с проблемами из-за недостаточного анализа существующих решений и неправильной оценки конкурентной среды.



Качественный анализ аналогов позволяет:

- **Избежать изобретения велосипеда** - понять, что уже реализовано на рынке
- **Выявить нишу** - определить неудовлетворенные потребности пользователей
- **Обосновать актуальность** - аргументировать необходимость разработки
- **Определить конкурентные преимущества** - сформулировать уникальное торговое предложение
- **Учесть опыт предшественников** - избежать чужих ошибок и применить лучшие практики
- **Оценить целесообразность разработки** - принять взвешенное решение о начале проекта



В контексте дипломного проекта данный раздел обязателен для большинства тем и обычно размещается в первой/второй главе работы под названием "Обзор существующих решений" и/или "Анализ аналогов".

## **Формулировка задания**

Проведите детальный сравнительный анализ существующих решений (систем-аналогов) в предметной области вашего дипломного проекта. На основе проведенного анализа обоснуйте актуальность и конкурентные преимущества разрабатываемого вами решения.

### **Необходимо выполнить:**

1. **Идентифицировать системы-аналоги**:
   - Выявите минимум 3-5 существующих решений, схожих с вашим проектом
   - Выберите как коммерческие, так и open-source решения (если применимо)
   - Обоснуйте выбор конкретных систем для анализа

2. **Провести функциональный анализ**:
   - Составьте сравнительную таблицу функциональных возможностей
   - Оцените полноту реализации функционала
   - Выявите отсутствующие или недостаточно реализованные функции

3. **Выполнить SWOT-анализ**:
   - Для каждого аналога определите сильные стороны (Strengths)
   - Выявите слабые стороны (Weaknesses)
   - Определите возможности (Opportunities)
   - Оцените угрозы (Threats)

4. **Провести технический анализ**:
   - Изучите используемые технологии и архитектуру
   - Оцените производительность и масштабируемость
   - Проанализируйте требования к инфраструктуре

5. **Выполнить UX/UI анализ** (для систем с интерфейсом):
   - Оцените удобство использования
   - Проанализируйте качество пользовательского опыта
   - Выявите проблемы интерфейса

6. **Сформулировать конкурентные преимущества**:
   - На основе выявленных недостатков аналогов определите уникальные возможности вашего решения
   - Обоснуйте, почему ваше решение будет лучше существующих
   - Определите целевую аудиторию, для которой ваше решение наиболее ценно

7. **Подготовить аналитическую записку**:
   - Резюмируйте результаты анализа
   - Сформулируйте выводы об актуальности вашего проекта
   - Обоснуйте целесообразность разработки

#### **Теоретическая часть:**

**Анализ аналогов (конкурентный анализ)** - это систематическое исследование существующих на рынке решений, схожих с разрабатываемым продуктом, с целью выявления их преимуществ, недостатков, функциональных возможностей и определения конкурентной позиции собственного проекта.

**Основные цели анализа аналогов:**
- Изучение состояния рынка и существующих решений
- Выявление неудовлетворенных потребностей пользователей
- Определение уникальных характеристик разрабатываемого продукта
- Формирование конкурентных преимуществ
- Обоснование актуальности проекта

### **2. Этапы проведения анализа**

#### **Этап 1. Идентификация систем-аналогов**

Прежде всего необходимо определить критерии отбора систем для анализа:

**Критерии отбора:**
- **Функциональная схожесть** - решение выполняет схожие задачи
- **Целевая аудитория** - ориентировано на ту же группу пользователей
- **Масштаб** - сопоставимый уровень сложности и охват
- **Доступность информации** - возможность получить данные для анализа
- **Актуальность** - решение активно развивается и используется





**Источники поиска аналогов:**
- Специализированные каталоги ПО (Capterra, G2, Software Advice)
- Репозитории open-source проектов (GitHub, GitLab)
- Научные публикации и патентные базы
- Обзоры в профессиональных изданиях
- Поисковые системы и маркетплейсы приложений

**Типы аналогов для анализа:**
- **Прямые конкуренты** - решения с идентичным функционалом
- **Косвенные конкуренты** - решения, закрывающие ту же потребность иным способом
- **Потенциальные конкуренты** - решения, которые могут выйти на ваш рынок
- **Продукты-заменители** - альтернативные способы решения проблемы

#### **Этап 2. Функциональный анализ**

**Сравнительная таблица функций** - основной инструмент функционального анализа.

**Структура таблицы:**
- Строки - функциональные возможности
- Столбцы - анализируемые системы
- Ячейки - наличие/отсутствие функции или оценка качества реализации

**Шкала оценки функционала:**
- **++** или **★★** - функция реализована полностью и качественно
- **+** или **★** - функция реализована частично или с ограничениями
- **-** или **✗** - функция отсутствует
- **?** - информация недоступна

**Группировка функций:**
Разделите функционал на логические группы для структурированного анализа:
- Основные функции (core features)
- Дополнительные функции (additional features)
- Административные функции (admin features)
- Интеграции
- Безопасность

#### **Этап 3. SWOT-анализ**

**SWOT-анализ** - стратегический метод анализа, оценивающий факторы внутренней и внешней среды организации или продукта.



**Компоненты SWOT:**

**S (Strengths - Сильные стороны)** - внутренние положительные характеристики:
- Уникальные функции
- Высокое качество
- Известность бренда
- Большая пользовательская база
- Техническое превосходство

**W (Weaknesses - Слабые стороны)** - внутренние отрицательные характеристики:
- Отсутствие важных функций
- Устаревшие технологии
- Проблемы с производительностью
- Сложный интерфейс
- Высокая стоимость

**O (Opportunities - Возможности)** - внешние благоприятные факторы:
- Растущий рынок
- Новые технологии
- Изменение потребностей пользователей
- Слабость конкурентов

**T (Threats - Угрозы)** - внешние неблагоприятные факторы:
- Появление новых конкурентов
- Изменение законодательства
- Технологические изменения
- Экономические факторы



**Практические рекомендации по SWOT-анализу:**
- Будьте объективны и реалистичны
- Используйте конкретные факты, а не общие фразы
- Ранжируйте факторы по значимости
- Анализируйте взаимосвязи между квадрантами
- Формулируйте стратегии на основе выявленных факторов

##### **5.2. ТЕОРЕТИЧЕСКИЕ ОСНОВЫ UX/UI ДИЗАЙНА**



###### **5.2.1. Закон Хика (Hick's Law, 1952) - Время принятия решения**

**Историческая справка:** Уильям Эдмунд Хик и Рэй Хайман независимо обнаружили логарифмическую зависимость между количеством альтернатив и временем принятия решения, основываясь на теории информации Шеннона.


**Математическая формула:**
$RT = a + b \cdot \log_2(n + 1)$

где:
- $RT$ - время реакции (reaction time) в миллисекундах
- $a$ - общее время, не связанное с принятием решения (≈200 мс для визуального стимула)
- $b$ - эмпирическая константа, зависящая от сложности решения (≈150 мс для простых решений)
- $n$ - количество альтернатив выбора

**Почему это работает:** Мозг обрабатывает альтернативы методом бинарного поиска, последовательно исключая половину вариантов. Количество информации, необходимой для выбора из n альтернатив, равно $\log_2(n)$ бит согласно теории информации.



**Практическое применение для анализа интерфейсов:**





- Для меню с 12 пунктами (Jira):

  - $RT_{Jira} = 200 + 150 \cdot \log_2(12 + 1) = 200 + 150 \cdot 3.7 = 755$ мс



- Для меню с 5 пунктами (Trello):

  - $RT_{Trello} = 200 + 150 \cdot \log_2(5 + 1) = 200 + 150 \cdot 2.58 = 587$ мс

**Вывод для комиссии:** "Меню Jira требует на 29% больше времени для принятия решения согласно закону Хика, что подтверждено эмпирическими исследованиями (Хик, 1952; Хайман, 1953)."

###### **5.2.2. Закон Фиттса (Fitts's Law, 1954) - Время достижения цели**


**Историческая справка:** Пол Фиттс разработал модель человеческого движения, основанную на теории информации, первоначально для военных целей (предсказание точности пилотов).



**Математическая формула:**
$MT = a + b \cdot \log_2\left(\frac{2D}{W}\right)$

где:
- $MT$ - время движения (movement time)
- $D$ - расстояние от начальной точки до центра цели
- $W$ - ширина цели (допустимая погрешность)
- $a, b$ - эмпирические константы (для мыши: $a$ ≈ 50 мс, $b$ ≈ 150 мс)



**Индекс сложности (Index of Difficulty):**
$ID = \log_2\left(\frac{2D}{W}\right)$

Измеряется в битах информации, необходимых для уточнения позиции.


**Почему это работает:** Движение к цели представляет собой серию субдвижений с визуальной обратной связью. Каждое субдвижение уменьшает расстояние до цели примерно вдвое (оптимальная стратегия управления).





**Стандарты минимальных размеров интерактивных элементов:**
- **iOS Human Interface Guidelines:** минимум 44×44 пикселя
- **Material Design (Android):** минимум 48×48 dp
- **Web Content Accessibility Guidelines (WCAG):** минимум 44×44 CSS пикселя


**Пример расчета для кнопки "Сохранить":**

- Плохой дизайн (маленькая кнопка 20×20 px):
  - $ID_{small} = \log_2\left(\frac{2 \cdot 500}{20}\right) = \log_2(50) = 5.64$ бит

- Хороший дизайн (стандартная кнопка 48×48 px):
  - $ID_{standard} = \log_2\left(\frac{2 \cdot 500}{48}\right) = \log_2(20.8) = 4.38$ бит





**Время нажатия:**
 - $MT_{small} = 50 + 150 \cdot 5.64 = 896$ мс
 - $MT_{standard} = 50 + 150 \cdot 4.38 = 707$ мс



**Вывод для комиссии:** "Кнопка размером 20px требует на 27% больше времени для точного нажатия согласно закону Фиттса, что увеличивает количество промахов и снижает эффективность работы."

###### **5.2.3. Закон Миллера (Miller's Law, 1956) - Магическое число 7±2**


**Историческая справка:** Джордж Миллер изучал пределы человеческой способности обрабатывать информацию и обнаружил, что емкость кратковременной памяти ограничена.



**Теоретическая основа - Емкость канала:**

- $C = \log_2 N$

где:
- $C$ - емкость канала в битах
- $N$ - количество различимых альтернатив



**Современные исследования (Cowan, 2001; Luck & Vogel, 1997):** Реальная емкость рабочей памяти составляет $4 \pm 1$ элемента для сложных объектов. Число 7±2 актуально только для простых, одномерных стимулов (цифры, буквы).

**Формула когнитивной перегрузки:**
$P_{error} = 1 - e^{-k(n-C)}$ при $n > C$

где:
- $P_{error}$ - вероятность ошибки
- $n$ - количество элементов
- $C$ - емкость памяти (≈4 для сложных элементов)
- $k$ - константа сложности задачи (≈0.5)

**Практический пример:**




- Меню с 12 пунктами (превышение на 8 элементов):
  - $P_{error} = 1 - e^{-0.5 \cdot (12-4)} = 1 - e^{-4} = 0.98$ (98% вероятность ошибки или необходимости повторного просмотра)

- Меню с 5 пунктами (превышение на 1 элемент):
  - $P_{error} = 1 - e^{-0.5 \cdot (5-4)} = 1 - e^{-0.5} = 0.39$ (39% вероятность)

**Стратегия чанкинга (группировки):** Информация группируется в смысловые блоки, каждый из которых занимает одну ячейку памяти.

Пример: телефонный номер
- Плохо: 89991234567 (11 элементов)
- Хорошо: 8 (999) 123-45-67 (4 чанка)









**Вывод для комиссии:** "Интерфейс с 12 элементами меню превышает когнитивную емкость на 200%, что приводит к 98% вероятности необходимости повторного сканирования согласно модели Коуэна."

###### **5.2.4. Теория визуального внимания (Feature Integration Theory, Treisman & Gelade, 1980)**


**Основа теории:** Визуальная обработка происходит в два этапа:
1. **Предвнимательная стадия** - параллельная обработка базовых признаков
2. **Стадия фокусированного внимания** - последовательное связывание признаков

**Время визуального поиска:**
$T = a + b \cdot n$ (для поиска с участием внимания)
$T = a$ (для предвнимательного поиска)

где:
- $T$ - время поиска
- $n$ - количество отвлекающих элементов
- $a$ - базовое время (≈400 мс)
- $b$ - время на элемент (≈50 мс для сложного поиска)





**Базовые визуальные признаки (обрабатываются параллельно):**
- Цвет
- Ориентация
- Размер
- Движение



**Практическое применение:** Важные элементы должны отличаться по базовым признакам для мгновенного обнаружения.



**Пример анализа:**
"Кнопка 'Сохранить' в Jira не выделяется цветом среди 8 других кнопок, что увеличивает время поиска: $T = 400 + 50 \cdot 8 = 800$ мс вместо 400 мс при цветовом выделении."

###### **5.2.5. Гештальт-принципы восприятия (1920-е годы)**


**Теоретическая основа:** Мозг стремится к организации визуальной информации в целостные структуры согласно принципам:


**1. Закон близости (Proximity)**
Метрика группировки - коэффициент близости:
$K_{proximity} = \frac{d_{within}}{d_{between}}$

где:
- $d_{within}$ - среднее расстояние внутри группы
- $d_{between}$ - среднее расстояние между группами

Оптимальное значение: $K_{proximity} < 0.5$ (расстояние внутри группы минимум в 2 раза меньше)

**2. Закон схожести (Similarity)**
Метрика схожести по цвету (CIE ΔE*2000):
$\Delta E = \sqrt{(\Delta L^*)^2 + (\Delta a^*)^2 + (\Delta b^*)^2}$

Интерпретация:
- $\Delta E < 1.0$ - неразличимо для человека
- $\Delta E < 2.3$ - едва заметная разница
- $\Delta E < 5.0$ - заметная разница
- $\Delta E > 5.0$ - явная разница







**3. Закон общей области (Common Region)**
Элементы внутри замкнутой области воспринимаются как группа. Эффективность группировки:
$E_{region} = 1 - \frac{P_{cross}}{P_{total}}$

где:
- $P_{cross}$ - периметр пересечения областей
- $P_{total}$ - общий периметр областей

**Практический пример анализа группировки:**
"В интерфейсе Jira отступы между элементами одной группы составляют 16px, между группами - 20px, что дает коэффициент $K_{proximity} = 16/20 = 0.8 > 0.5$, следовательно, группировка визуально неочевидна."

##### **5.3. КОЛИЧЕСТВЕННЫЕ МЕТРИКИ UX**


###### **Метрики по стандарту ISO 9241-11:2018**



**Определение юзабилити:**
$Usability = f(Effectiveness, Efficiency, Satisfaction, Context)$


**1. Эффективность (Effectiveness):**
$Effectiveness = \frac{\sum_{i=1}^{N} \sum_{j=1}^{M} n_{ij} \times w_j}{N \times \sum_{j=1}^{M} w_j} \times 100\%$

где:
- $n_{ij}$ - результат задачи $j$ для пользователя $i$ (0 или 1)
- $w_j$ - вес задачи $j$
- $N$ - количество пользователей
- $M$ - количество задач

**2. Эффективность (Efficiency):**




$Time Efficiency = \frac{1}{N} \sum_{i=1}^{N} \sum_{j=1}^{M} \frac{n_{ij}}{t_{ij}}$



где $t_{ij}$ - время выполнения задачи $j$ пользователем $i$

**3. Относительная эффективность пользователя:**


$Relative User Efficiency = \frac{\sum_{i=1}^{N} \sum_{j=1}^{M} n_{ij} \times t_{ij}^{expert}}{\sum_{i=1}^{N} \sum_{j=1}^{M} n_{ij} \times t_{ij}^{user}} \times 100\%$


**Пример расчета для Jira:**
- Задача: создать и назначить задачу
- Время эксперта: 20 секунд
- Среднее время пользователя: 45 секунд
- Успешность: 78%

- $Effectiveness = 0.78 \times 100\% = 78\%$


- $Efficiency = \frac{0.78}{45} = 0.0173$ задач/сек


- $Relative Efficiency = \frac{20}{45} \times 100\% = 44.4\%$


**Вывод:** "Эффективность Jira составляет 44.4% от эталонной, что указывает на проблемы с юзабилити согласно ISO 9241-11."


##### **5.4. ОБЪЕКТИВНЫЕ МЕТРИКИ ВИЗУАЛЬНОГО ДИЗАЙНА**


###### **A. Визуальная иерархия и контраст**


**1. Формула контрастности WCAG 2.1:**


$CR = \frac{L_1 + 0.05}{L_2 + 0.05}$

где $L_1$ - относительная яркость более светлого цвета, $L_2$ - более темного

**Расчет относительной яркости:**
$L = 0.2126 \times R_{linear} + 0.7152 \times G_{linear} + 0.0722 \times B_{linear}$

**Линеаризация sRGB:**
$C_{linear} = \begin{cases}
\frac{C_{sRGB}}{12.92} & \text{если } C_{sRGB} \leq 0.03928 \\
\left(\frac{C_{sRGB} + 0.055}{1.055}\right)^{2.4} & \text{иначе}
\end{cases}$

**Требования WCAG 2.1:**
- Обычный текст: $CR \geq 4.5:1$
- Крупный текст (14pt жирный или 18pt обычный): $CR \geq 3:1$
- Уровень AAA: $CR \geq 7:1$

**2. Визуальный вес элемента:**
$W_{visual} = S \times C \times P \times M$

где:
- $S$ - относительный размер (площадь элемента / площадь экрана)
- $C$ - контраст с фоном (нормализованный 0-1)
- $P$ - позиционный вес (центр = 1, периферия = 0.6)
- $M$ - движение (статичный = 1, анимированный = 1.5)

###### **B. Типографические метрики**


**1. Оптимальная длина строки (Dyson & Haselgrove, 2001):**



Скорость чтения в зависимости от длины строки:
$Reading Speed = -0.002 \times (CPL - 66)^2 + 250$

где CPL - количество символов в строке

Максимум при CPL = 66 символов (250 слов/мин)

**2. Индекс читабельности Флеша-Кинкейда:**



$FKGL = 0.39 \times \frac{words}{sentences} + 11.8 \times \frac{syllables}{words} - 15.59$

Интерпретация:
- FKGL < 8: легко читается большинством
- FKGL 8-12: средняя сложность
- FKGL > 12: сложный текст





**3. Межстрочный интервал:**



$Line Height = Font Size \times k$

где $k$ оптимально:
- Для основного текста: $1.5 \leq k \leq 1.7$
- Для заголовков: $1.2 \leq k \leq 1.3$
- Для мобильных устройств: $k \geq 1.6$



##### **5.5. КОГНИТИВНАЯ НАГРУЗКА - ИЗМЕРЕНИЕ**



###### **Модель когнитивной нагрузки (Sweller, 1988)**

**Общая когнитивная нагрузка:**
$CL_{total} = CL_{intrinsic} + CL_{extraneous} + CL_{germane}$

При этом должно выполняться:
$CL_{total} \leq WM_{capacity}$

где $WM_{capacity}$ ≈ 4 единицы (Cowan, 2001)

**Измерение когнитивной нагрузки - NASA-TLX:**



$TLX = \frac{\sum_{i=1}^{6} rating_i \times weight_i}{\sum_{i=1}^{6} weight_i}$

Шкалы оценки (0-100):
1. Mental Demand (умственная нагрузка)
2. Physical Demand (физическая нагрузка)
3. Temporal Demand (временная нагрузка)
4. Performance (производительность)
5. Effort (усилия)
6. Frustration (фрустрация)


**Индекс визуальной сложности (Rosenholtz et al., 2007):**
$VC = \alpha \cdot FC + \beta \cdot SC$

где:
- $FC$ - feature congestion (перегруженность признаками)
- $SC$ - subband entropy (энтропия поддиапазонов)
- $\alpha, \beta$ - весовые коэффициенты

**Расчет перегруженности признаками:**
$FC = \sigma_{color} + \sigma_{luminance} + \sigma_{orientation}$

##### **5.6. МЕТРИКИ ОШИБОК И ВОССТАНОВЛЕНИЯ**



###### **Классификация ошибок (Norman, 1983)**


**1. Промахи (Slips):** правильное намерение, неправильное выполнение

**2. Ляпсусы (Lapses):** ошибки памяти

**3. Ошибки (Mistakes):** неправильное намерение







**Метрика частоты ошибок:**

- $Error Rate = \frac{Total Errors}{Total Attempts} \times 100\%$

**Метрика восстановления после ошибок:**

- $Recovery Rate = \frac{Recovered Errors}{Total Errors} \times 100\%$

**Время восстановления:**
- $Recovery Time = t_{complete} - t_{error} - t_{normal}$

##### **5.7. ЭМОЦИОНАЛЬНЫЙ ДИЗАЙН - ИЗМЕРЕНИЕ**



###### **Модель эмоций PAD (Mehrabian & Russell, 1974)**

Эмоциональное состояние описывается вектором:
$Emotion = (P, A, D)$

где:
- $P$ - Pleasure (удовольствие): от -1 до +1
- $A$ - Arousal (возбуждение): от -1 до +1
- $D$ - Dominance (доминирование): от -1 до +1

**Целевые значения для продуктивных приложений:**
- $P = 0.6$ (умеренно приятный)
- $A = 0.3$ (слегка стимулирующий)
- $D = 0.7$ (высокий контроль пользователя)

###### **Эстетическая привлекательность (Ngo et al., 2003)**


**Метрика визуального баланса:**

- $Balance = 1 - \frac{|W_L - W_R|}{W_L + W_R}$

где $W_L$, $W_R$ - визуальные веса левой и правой частей



**Метрика симметрии:**

- $Symmetry = 1 - \frac{\sum|p_i - p_i'|}{n \times max\_distance}$

где $p_i$, $p_i'$ - соответствующие точки относительно оси симметрии

##### **5.8. СТАТИСТИЧЕСКАЯ ВАЛИДАЦИЯ РЕЗУЛЬТАТОВ**



###### **Определение размера выборки для юзабилити-тестирования**


**Формула Nielsen (2000) для обнаружения проблем:**
$P = 1 - (1 - p)^n$

где:
- $P$ - вероятность обнаружения проблемы
- $p$ - доля пользователей с проблемой (≈0.31 для типичных интерфейсов)
- $n$ - количество участников

Для 85% проблем: $n = 5$
Для 95% проблем: $n = 10$


###### **Статистическая значимость различий**



**t-тест для сравнения времени выполнения задач:**
$t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$

**Размер эффекта Коэна:**
$d = \frac{\bar{x}_1 - \bar{x}_2}{s_{pooled}}$

Интерпретация:
- $d < 0.2$ - незначительный эффект
- $0.2 \leq d < 0.5$ - малый эффект
- $0.5 \leq d < 0.8$ - средний эффект
- $d \geq 0.8$ - большой эффект

##### **5.9. ПРИМЕР КОМПЛЕКСНОГО АНАЛИЗА ДЛЯ ЗАЩИТЫ**


**Анализ системы Jira по объективным метрикам:**




**1. Закон Хика - время выбора в меню:**
- Количество пунктов: 12
- Расчетное время: $RT = 200 + 150 \times \log_2(13) = 761$ мс
- Оптимальное время (5 пунктов): 587 мс

**2. Закон Фиттса - размер кнопок:**
- Размер кнопки "Создать": 60×32 пикселей
- Индекс сложности: $ID = \log_2(2 \times 400/32) = 4.64$ бит
- Стандарт WCAG (44×44): $ID = 3.92$ бит
- **Вывод:** увеличение времени клика на 18%

**3. Когнитивная нагрузка:**
- Элементов на главном экране: 47
- Превышение нормы Миллера: 47/7 = 6.7 раз
- NASA-TLX оценка: 72/100 (высокая нагрузка)
- **Вывод:** требуется упрощение интерфейса

**4. Контрастность (WCAG 2.1):**
- Основной текст: #5e6c84 на #ffffff
- Контрастность: 4.52:1
- **Вывод:** соответствует минимальным требованиям AA

**5. Эффективность по ISO 9241-11:**
- Task Completion Rate: 78%
- Относительная эффективность: 44%
- **Вывод:** низкая эффективность использования

##### **5.11. Контрольный чеклист**






Перед защитой убедитесь, что для каждой проблемы UX/UI у вас есть:

- **Конкретная метрика или формула** (закон Хика, Фиттса, и т.д.)
- **Численное значение** (время в мс, проценты, баллы)
- **Сравнение с эталоном или стандартом** (WCAG, ISO, HIG)
- **Ссылка на научное исследование** (автор, год)
- **Расчет или измерение** (формула с подставленными значениями)
- **Интерпретация результата** (что означает для пользователя)
- **Визуальное доказательство** (скриншот, схема, график)
- **Статистическая значимость** (при наличии выборки)



Это позволит вам уверенно обосновать свои выводы перед комиссией и доказать, что ваш анализ базируется на научных принципах и объективных измерениях, а не на субъективных предпочтениях.

- **Эвристическая оценка** - проверка по принципам юзабилити
- **Когнитивное прохождение** - имитация действий пользователя
- **A/B тестирование** - сравнение вариантов интерфейса
- **Анализ метрик** - изучение данных об использовании
- **Обзор отзывов пользователей** - анализ обратной связи





**Инструменты для UX/UI анализа:**
- Скриншоты интерфейсов
- Карты путей пользователя (User Journey Maps)
- Аннотированные wireframes
- Видеозаписи взаимодействия

#### **Этап 6. Формулирование конкурентных преимуществ**

**Конкурентное преимущество** - характеристика продукта, отличающая его от конкурентов и создающая дополнительную ценность для пользователей.


**Виды конкурентных преимуществ:**



-  **1. Функциональные:**
      - Уникальные возможности
      - Более полный функционал
      - Лучшая реализация стандартных функций

-  **2. Технологические:**
    - Использование передовых технологий
    - Лучшая производительность
    - Более высокая надежность

-  **3. Пользовательские:**
    - Более удобный интерфейс
    - Лучший пользовательский опыт
    - Персонализация

-  **4. Экономические:**
    - Более низкая цена
    - Лучшее соотношение цена/качество
    - Гибкая модель монетизации

-  **5. Стратегические:**
    - Фокус на специфичной нише
    - Адаптация под особые потребности
    - Локализация



**Формулировка уникального торгового предложения (УТП):**

- УТП должно отвечать на три вопроса:
  1. Что вы предлагаете?
  2. Чем это отличается от конкурентов?
  3. Почему это важно для пользователя?



**Формула УТП:**



```
[Продукт] помогает [целевой аудитории] [решить проблему/получить выгоду],
в отличие от [конкуренты], благодаря [уникальная характеристика].
```

#### **5. Выявленные проблемы существующих решений**

На основе проведенного анализа выявлены следующие ключевые проблемы:

**1. Дихотомия сложности:**
- Jira: избыточная сложность для малых и средних команд
- Trello/GitHub Projects: недостаточный функционал для полноценного управления проектами
- **Проблема:** отсутствие решения с оптимальным балансом простоты и функциональности

**2. Зависимость от интернета:**
- Все рассмотренные коммерческие решения требуют постоянного онлайн-подключения
- **Проблема:** невозможность работы в условиях нестабильного интернета или повышенных требований к безопасности

**3. Слабая интеграция с инструментами тестирования:**
- Существующие системы не имеют глубокой интеграции с frameworks тестирования
- **Проблема:** разрозненность процессов разработки и тестирования

**4. Стоимость:**
- Полнофункциональные решения (Jira, Asana) дороги для малых команд
- Бесплатные решения имеют существенные ограничения
- **Проблема:** высокий порог входа для стартапов и малых команд

**5. Кастомизация:**
- Коммерческие решения имеют ограничения в кастомизации
- Open-source решения требуют высоких технических компетенций
- **Проблема:** сложность адаптации под специфические потребности команды

#### **6. Конкурентные преимущества разрабатываемого решения**

На основе выявленных проблем определены следующие конкурентные преимущества нашей системы:

- Стоимость платной версии $5-8 за пользователя/месяц
- **Ценность:** доступность для стартапов и малых команд

**5. Современный UX/UI:**
- Дизайн, соответствующий современным трендам

- Темная тема и настройка визуала

- Учет специфики agile-методологий в разработке

### 4.1. Environmental Assessment Studio

#### 4.1.1. Общее описание

QGIS-плагин с открытым исходным кодом, разработан независимым разработчиком (V. A. Anaba), интегрирует Google Earth Engine непосредственно в QGIS. Актуальная версия 4.5.1, распространяется через официальный репозиторий плагинов QGIS и GitHub.

#### 4.1.2. Основные функциональные возможности

Более 20 видов экологического анализа: землепользование/земельный покров, change detection, температура поверхности и городские тепловые острова, анализ затоплений, вегетационные индексы (NDVI, NDWI), деградация земель и засуха, органический углерод почв, риски пожаров и эрозии, качество воздуха, фрагментация местообитаний. Пользователь задаёт область интереса и период дат, плагин обрабатывает снимки Sentinel/Landsat через Earth Engine.

#### 4.1.3. Технические характеристики

На выходе — стилизованные растры, обработанные стеки снимков, векторные слои, таблицы площадей по классам, файлы метаданных, экспорт трендового анализа. Требует настроенного аккаунта Google Earth Engine.

#### 4.1.4. Преимущества

- Самый широкий охват видов анализа среди рассмотренных решений
- Облачные вычисления Earth Engine избавляют от ручного скачивания снимков
- Открытый код

#### 4.1.5. Недостатки

- Инструмент общего назначения, не заточен под задачу «объект застройки × лес»
- Требует аккаунта GEE — барьер входа
- Нет функции пространственного пересечения с объектами и нет прогнозирования
- Интерфейс на английском.

### 4.2. URGREECS (Urban Green Equity Analytics 3-30-300)

#### 4.2.1. Общее описание

Открытый QGIS-плагин (F. Afrianto, M. Safira), реализует международную концепцию «3-30-300» для оценки равенства доступа к городской зелени. Версия 1.0.0, GitHub.

#### 4.2.2. Основные функциональные возможности


Анализ на уровне здания (крона рядом), квартала (доля зелени ≥30%) и доступности (парк в радиусе 300 м); отдельно учитывает уязвимые группы населения.

Входные данные: модель высоты крон, растры населения и уязвимых групп, контуры зданий, границы кварталов, слои общественной зелени, административные границы. На выходе — показатели по зданиям/кварталам, оценки «зелёного равенства», приоритетные классификации, дашборды, CSV.

#### 4.2.4. Преимущества

- Единственный из аналогов, учитывающий социальную составляющую (уязвимые группы)
- Признанная международная методика
- Готовые визуализации и отчётность

#### 4.2.5. Недостатки

- Оценивает доступ жителей к зелени, а не воздействие конкретного проекта застройки на лес
- Не работает с временными рядами и не прогнозирует
- Высокий порог входа по подготовке исходных слоёв

- Нет пространственной визуализации воздействия

### 4.4. i-Tree Canopy

#### 4.4.1. Общее описание

Бесплатный веб-инструмент, поддерживается партнёрством во главе с USDA Forest Service с 2006 года, доступен на canopy.itreetools.org.

#### 4.4.2. Основные функциональные возможности

Классификация земельного/древесного покрова методом случайной выборки точек по аэрофотоснимкам Google Maps: пользователь вручную классифицирует 500–1000 точек, на основе чего рассчитывается доля кроны и экологические выгоды (секвестрация CO₂, снижение загрязнения воздуха, регулирование стока).

#### 4.4.3. Технические характеристики

Границы проекта — через готовые геоданные, ручное рисование или шейп-файлы; поддержка нескольких участков одновременно; требователен к трафику.

#### 4.4.4. Преимущества

- Давно используется в отрасли, доступен бесплатно и не требует знаний программирования
- Даёт не только площадь, но и денежно-экологическую оценку выгод от кроны


#### 4.4.5. Недостатки

- Ручная классификация требует много времени, особенно при работе с большими объёмами данных, и зависит от субъективной оценки специалиста
- Оценка проводится только на текущий момент, без возможности отслеживать изменения и строить прогнозы
- Отсутствует интеграция с ГИС
- Финансирование со стороны USDA недавно приостановлено, поэтому дальнейшая поддержка инструмента остаётся под вопросом.

### 4.5. Методика №897-ПП

#### 4.5.1. Общее описание

Действующая официальная методика оценки размера вреда от повреждения/уничтожения зелёных насаждений в Москве (Постановление Правительства Москвы от 14.11.2006), применяется совместно с порядком компенсационного озеленения (№159-ПП).

#### 4.5.2. Основные функциональные возможности

Формализованный расчёт денежной компенсации за уничтоженные насаждения на основе таксационных показателей (порода, возраст, диаметр ствола), зафиксированных актом обследования.

#### 4.5.3. Технические характеристики

Выполняется вручную специалистами-таксаторами по акту осмотра; не предполагает пространственного анализа, снимков или автоматизации.

#### 4.5.4. Преимущества

- Официально утверждена и обязательна к применению в Москве
- Учитывает особенности зелёного фонда Москвы и действующие правовые нормы

#### 4.5.5. Недостатки

- Применяется после фактического повреждения зелёных насаждений и не позволяет прогнозировать возможный ущерб
- Не предусматривает использование пространственных данных и методов дистанционного зондирования
- Требует значительного объёма ручной работы
- Не позволяет оценивать потенциальное воздействие конкретных градостроительных проектов на этапе их планирования

## 8. ВЫЯВЛЕННЫЕ ПРОБЛЕМЫ И ОГРАНИЧЕНИЯ СУЩЕСТВУЮЩИХ РЕШЕНИЙ

### 8.1. Отсутствие привязки анализа к конкретному проекту застройки

**Описание:** Ни один из рассмотренных инструментов не выполняет прямое пространственное сопоставление границ конкретного строительного/реновационного проекта с конкретными лесными участками. EAS, URGREECS и QGYF работают с обобщёнными показателями территории (весь город/квартал/произвольная область), i-Tree Canopy — с произвольно нарисованной областью без привязки к проекту застройки, а 897-ПП применяется постфактум к уже зафиксированному факту повреждения, а не к запланированному проекту.

**Влияние:** Специалистам приходится вручную, без инструментальной поддержки, сопоставлять границы объектов программы реновации с картой лесов — трудоёмко, немасштабируемо при большом числе объектов программы, невоспроизводимо (разные специалисты могут получить разные оценки одного и того же объекта).

**Распространенность:** Во всех пяти рассмотренных решениях.

### 8.2. Отсутствие прогнозной составляющей

**Описание:** Ни один инструмент не анализирует динамику состояния лесного покрова во времени для построения прогноза будущего воздействия. EAS технически умеет делать change detection, но не применяет его для прогнозирования; остальные либо статичны (оценка на один момент времени), либо ретроспективны (897-ПП фиксирует уже случившийся ущерб).

**Влияние:** Оценка воздействия возможна только по факту — после начала или завершения работ, что не позволяет предотвратить ущерб на этапе планирования и скорректировать проект заранее.

**Распространенность:** Во всех пяти решениях.

### 8.3. Разрыв между автоматизированным ГИС-анализом и официальной нормативной процедурой

**Описание:** Инструменты с автоматизированным геопространственным анализом (EAS, URGREECS, QGYF) существуют отдельно от официальной методики оценки ущерба (897-ПП) — между ними нет связи; официальная методика выполняется вручную, без использования ГИС или данных дистанционного зондирования, хотя технически это уже возможно.

**Влияние:** Городские специалисты вынуждены вести два несвязанных процесса — таксационное обследование по регламенту и, отдельно, любой ГИС-анализ (если он вообще проводится) — результаты не соотносятся друг с другом напрямую.

**Распространенность:** На стыке категорий «готовый софт» и «официальная методика» — касается всех шести рассмотренных позиций сразу.

### 8.4. Отсутствие локализации под российский контекст

**Описание:** Все четыре программных аналога (EAS, URGREECS, QGYF, i-Tree Canopy) разработаны и ориентированы на зарубежный контекст, интерфейсы на английском языке, без учёта специфики российского законодательства о зелёных насаждениях (категории ООПТ, нормативы, принятые в московском регулировании).

**Влияние:** Для использования в российском/московском контексте потребовалась бы дополнительная адаптация или ручной перевод результатов в термины, принятые в местном регулировании.

**Распространенность:** Во всех четырёх программных аналогах (не относится к 897-ПП — это и есть локальный документ).

## 9. КОНКУРЕНТНЫЕ ПРЕИМУЩЕСТВА РАЗРАБАТЫВАЕМОГО РЕШЕНИЯ

### 9.1. Прямая привязка анализа к конкретному объекту программы реновации

**Описание:** Инструмент выполняет пространственное сопоставление границ конкретного объекта реновации с конкретными лесными участками (пересечение + буферная зона), а не работает с обобщёнными показателями территории.

**Обоснование:** Решает проблему 8.1 — ни у одного аналога нет привязки анализа к конкретному проекту застройки.

**Ценность для пользователя:** Специалист получает готовый, воспроизводимый расчёт по каждому объекту программы, без ручного сопоставления слоёв вручную.

**Отличие от конкурентов:** EAS/URGREECS/QGYF анализируют территорию в целом, i-Tree Canopy — произвольную область без учёта конкретного проекта, 897-ПП фиксирует уже случившееся повреждение, а не работает на этапе планирования конкретного объекта.

### 9.2. Прогнозная оценка на основе временной динамики

**Описание:** Анализ временного ряда спутниковых снимков за период, предшествующий реализации проекта, для построения прогноза ожидаемого воздействия — до начала работ, а не после.

**Обоснование:** Решает проблему 8.2 — отсутствие прогнозной составляющей у всех рассмотренных решений.

**Ценность для пользователя:** Воздействие можно учитывать при принятии решений на этапе планирования, а не фиксировать ущерб постфактум.

**Отличие от конкурентов:** EAS технически умеет делать change detection, но не применяет его для прогноза; остальные либо статичны (оценка на момент времени), либо ретроспективны (897-ПП).

### 9.3. Связующее звено между ГИС-анализом и официальной методикой

**Описание:** Инструмент проектируется как дополнение к официальной методике №897-ПП: даёт количественную пространственную оценку на этапе проектирования, результаты которой могут использоваться как основание при последующем применении официальной методики расчёта ущерба.

**Обоснование:** Решает проблему 8.3 — разрыв между автоматизированным анализом и нормативной процедурой.

**Ценность для пользователя:** Органы власти получают единую цепочку «прогноз → официальный расчёт» вместо двух несвязанных процессов.

**Отличие от конкурентов:** Ни один из готовых программных инструментов не связан с 897-ПП, а сама методика не использует ГИС или данные дистанционного зондирования.

### 9.4. Локализация

**Описание:** Интерфейс и логика инструмента ориентированы на российское законодательство о зелёных насаждениях (категории ООПТ, нормативы, принятые в Москве), с русскоязычным интерфейсом.

**Обоснование:** Решает проблему 8.4 — отсутствие локализации у зарубежных аналогов.

**Ценность для пользователя:** Не требуется адаптация терминологии и подходов под местное регулирование — инструмент сразу применим в реальной практике.

**Отличие от конкурентов:** EAS, URGREECS, QGYF, i-Tree Canopy — англоязычные, разработаны без учёта российской нормативной базы.

## 10. ЦЕЛЕВАЯ АУДИТОРИЯ

### 10.1. Первичная целевая аудитория

**Описание:** Специалисты, непосредственно готовящие или принимающие решения по конкретным объектам программы реновации: проектировщики объектов реновации, специалисты Департамента природопользования и охраны окружающей среды г. Москвы (либо иного органа экологического контроля), а также специалисты Мосгоргеотреста, работающие с пространственными данными по этим объектам.


**Характеристики:**
- Работают в среде QGIS или имеют доступ к специалистам, которые в ней работают
- Принимают решения по конкретным, локализованным объектам, а не по городу в целом
- Нуждаются в воспроизводимом, документируемом результате для обоснования решений перед руководством, застройщиком или надзорными органами

**Потребности:**
- Быстро получить количественную оценку воздействия конкретного объекта на лес — без привлечения стороннего анализа под каждый случай
- Иметь картографический и табличный результат, пригодный для включения в проектную/надзорную документацию
- Видеть не только текущее состояние, но и прогноз — чтобы влиять на решение до, а не после реализации объекта


### 10.2. Вторичная целевая аудитория

**Описание:** Экологические НКО, независимые эксперты и заинтересованная общественность, а также исследователи и студенты, занимающиеся смежными темами городской экологии.

**Характеристики:**
- Не обязательно владеют QGIS или ГИС-инструментами напрямую — чаще потребляют готовые результаты (карты, отчёты), чем сами запускают анализ
- Заинтересованы в независимой проверке официальных оценок воздействия
- Действуют в публичном поле — используют результаты для общественного обсуждения, а не для принятия административных решений

**Потребности:**
- Возможность самостоятельно воспроизвести расчёт на тех же исходных данных, чтобы проверить официальные оценки
- Понятное представление результата (карта, а не только таблица показателей) для использования в публичной коммуникации

## 11. ВЫВОДЫ

### 11.1. Общие выводы по анализу рынка

Анализ пяти решений (три QGIS-плагина, один веб-инструмент, одна официальная методика) показал сегментированный рынок: EAS даёт широкий, но не заточенный под задачу технический инструментарий работы со снимками; URGREECS и QGYF решают смежные задачи — доступность и нормирование зелени, а не оценку воздействия застройки; i-Tree Canopy — признанная, но трудоёмкая ручная методика; 897-ПП формализует расчёт ущерба, но действует постфактум и без пространственного анализа. Ни одно решение не сочетает привязку к конкретному объекту застройки, анализ временной динамики и прогнозирование в одном инструменте.

### 11.2. Обоснование актуальности проекта

Программа реновации затрагивает территории, часть которых граничит с лесными массивами или включает их, а систематической, воспроизводимой и прогнозной оценки этого воздействия в существующей практике нет — она либо выполняется вручную постфактум (897-ПП), либо доступна только в инструментах общего назначения, требующих адаптации под задачу. Специализированный инструмент закрывает эту нехватку и позволяет учитывать воздействие на лес уже на этапе планирования объектов.

### 11.3. Позиционирование разрабатываемого решения

Инструмент занимает нишу на стыке автоматизированного геопространственного анализа (как у EAS/URGREECS/QGYF) и институционального нормативного контекста (как у 897-ПП): не заменяет официальную методику оценки ущерба, а работает на более раннем, прогнозном этапе — до момента, когда возникнет необходимость применения 897-ПП. Технически ближе всего к EAS (по глубине работы со снимками), но специализирован под задачу и локализован под российский контекст, чего нет ни у одного аналога.

### 11.4. Риски и ограничения

- Зависимость от доступности и качества исходных данных: точность прогноза напрямую зависит от доступности актуальных данных о границах объектов реновации, лесных территорий и качественных разновременных снимков — состав и доступность этих источников на данный момент не подтверждены (см. Задание №1, п.4) и требуют уточнения на следующих этапах.
- Методическая неопределённость самого прогноза: экстраполяция тренда по историческим данным требует научно обоснованной модели — это выходит за рамки чисто инженерной части и требует проработки с научным руководителем, иначе прогноз рискует быть неточным или неверифицируемым.
- Отсутствие апробации на реальных данных: инструмент разрабатывается впервые, без предварительного пилотного теста на конкретном объекте программы — на защите это может стать предметом вопросов о практической применимости.
- Инструмент не делает денежную оценку ущерба — осознанное ограничение объёма работы, а не недостаток по сравнению с 897-ПП: расчёт вреда остаётся отдельным, уже существующим процессом.
