# **Задание №1. Определение концептуальных основ дипломного проекта в области информационных систем**



В рамках дисциплины "Управление проектами информационных систем" представьте развернутое концептуальное обоснование вашего проекта, ответив на следующие вопросы:


#### **1. Обоснование необходимости разработки**


Для кого предназначено разрабатываемое Вами решение и какие потребности целевой аудитории оно призвано удовлетворить?




*Обоснуйте актуальность решаемой проблемы для конкретных заинтересованных сторон.*

<font color='red'><b>Ваш ответ</b></font>

Разрабатываемое решение предназначено для специалистов в области дистанционного зондирования Земли, геоинформационных систем и анализа пространственных данных, а также для организаций, занимающихся мониторингом окружающей среды, сельского хозяйства и урбанизированных территорий. К целевой аудитории относятся исследователи, аналитики данных, государственные структуры и коммерческие компании, использующие спутниковые снимки для принятия управленческих решений.

Современные методы анализа спутниковых изображений активно используют технологии глубокого обучения, однако большинство существующих моделей, включая ConvNeXt, ориентированы на обработку стандартных RGB-изображений. При этом реальные спутниковые данные являются мультиспектральными и содержат значительно большее количество каналов, отражающих различные физические характеристики объектов.

#### **2. Ожидаемые результаты проекта**



Опишите конкретные результаты, которые планируется получить по завершении проекта. Какой программный продукт, методика или информационная система будет создан/создана в результате вашей работы?

<font color='red'><b>Ваш ответ</b></font>

В результате выполнения дипломного проекта планируется разработка программного решения для анализа мультиспектральных спутниковых изображений на основе адаптированной архитектуры ConvNeXt. Основным результатом работы станет модифицированная модель глубокого обучения, способная эффективно обрабатывать многоканальные данные дистанционного зондирования Земли и использовать дополнительную спектральную информацию.В рамках проекта будет реализован программный прототип, включающий этапы загрузки и предобработки мультиспектральных данных, обучения модели и получения результатов классификации. Дополнительно планируется проведение сравнения базовой версии модели (с использованием RGB-данных) и модифицированной модели.

#### **3. Методология использования**


Каким образом предполагается практическое применение разрабатываемого решения? Опишите основные сценарии взаимодействия пользователей с разрабатываемым решением.

<font color='red'><b>Ваш ответ</b></font>

Предполагается следующий базовый сценарий взаимодействия пользователя с системой. Пользователь загружает мультиспектральные спутниковые изображения (например, данные с Sentinel-2) в систему, после чего выполняется их предварительная обработка, включая нормализацию и выбор спектральных каналов. Далее изображения передаются в обученную модель, которая выполняет анализ и классификацию, используя как пространственные, так и спектральные признаки.

Результатом работы системы является набор классифицированных данных, который может быть представлен в виде карт распределения классов (например, типов земного покрова), а также численных показателей качества.

#### **4. Входные данные и их источники**


Какие типы данных будут использоваться в качестве входных данных для вашего решения? В каком формате они будут представлены и из каких источников планируется их получение?

<font color='red'><b>Ваш ответ</b></font>

В качестве входных данных для разрабатываемого решения будут использоваться мультиспектральные спутниковые изображения, полученные в рамках программ дистанционного зондирования Земли.

Данные будут представлены в формате растровых изображений (GeoTIFF), содержащих многоканальную информацию. Каждый канал соответствует определённому спектральному диапазону, что делает возможным более детальный анализ по сравнению с обычными RGB-изображениями.

Основным источником данных являются спутники Sentinel-2, предоставляющие открытые мультиспектральные данные через платформу Copernicus Open Access Hub. Дополнительно планируется использование готовых размеченных датасетов, таких как EuroSAT и BigEarthNet, которые содержат аннотированные изображения и позволяют проводить обучение и тестирование моделей.

Какие результаты будет формировать Ваше решение на выходе?

<font color='red'><b>Ваш ответ</b></font>

В результате работы разрабатываемого решения формируются классифицированные спутниковые изображения, представленные в виде карт распределения классов (например, типы земного покрова: растительность, вода, застройка, сельскохозяйственные угодья). Также на выходе предоставляются численные метрики качества модели, позволяющие оценить эффективность обработки данных.

Использование разработанного решения снижает необходимость ручной обработки данных, ускоряет анализ и повышает качество принимаемых решений за счёт более полного использования доступной информации.
